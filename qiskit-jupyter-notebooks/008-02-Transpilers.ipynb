```python
circ = QuantumCircuit(3)
circ.ccx(0, 1, 2)
circ.draw()

pass_ = BasisTranslator(SessionEquivalenceLibrary, ['u1', 'u2', 'u3', 'cx'])  # Unroller was removed
pm = PassManager(pass_)
new_circ = pm.run(circ)
new_circ.draw(output='mpl')

try:
    pass_manager_drawer(pm, "passmanager.jpg")
except Exception as e:  # needs the Graphviz binaries installed
    print(e)
```
