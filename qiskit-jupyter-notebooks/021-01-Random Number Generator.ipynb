```python
# Random Number Generator
#Demonstrates the generation of an integer random number (from 0 to 31) on IBM Q Experience

%matplotlib inline
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit
from qiskit_backend_config import execute, Aer, IBMQ
from qiskit.compiler import transpile
from qiskit_backend_config import assemble
# from qiskit.tools.jupyter import *  # qiskit.tools.jupyter was removed in Qiskit 1.0
from qiskit.visualization import *
provider = IBMQ.load_account()
```

```python
q = QuantumRegister(5, 'q')
c = ClassicalRegister(5, 'c')
circuit = QuantumCircuit(q, c)

circuit.h(q[0])
circuit.h(q[1])
circuit.h(q[2])
circuit.h(q[3])
circuit.h(q[4])

circuit.measure(q, c)
```

```python
%matplotlib inline
circuit.draw(output="mpl")
```

```python
#backend = IBMQ.get_provider().get_backend('ibmq_london') 
backend = Aer.get_backend('qasm_simulator')
job = execute(circuit, backend=backend, shots=8192)
result = job.result()

counts = result.get_counts(circuit)
from qiskit.visualization import plot_histogram
plot_histogram(counts)
```
